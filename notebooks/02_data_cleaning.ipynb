```python
df_clean = df.copy()
```

```python
missing_percentage = (df_clean.isnull().sum() / len(df_clean)) * 100

print("Missing Values Percentage:")
print(missing_percentage[missing_percentage > 0])

```

```python
# Group by degree specialization and fill academic scores
for degree in df_clean['degree_specialization'].unique():
    mask = (df_clean['degree_specialization'] == degree) & (df_clean['ssc_percentage'].isnull())
    median_ssc = df_clean.loc[df_clean['degree_specialization'] == degree, 'ssc_percentage'].median()
    df_clean.loc[mask, 'ssc_percentage'] = median_ssc
    
    mask = (df_clean['degree_specialization'] == degree) & (df_clean['hsc_percentage'].isnull())
    median_hsc = df_clean.loc[df_clean['degree_specialization'] == degree, 'hsc_percentage'].median()
    df_clean.loc[mask, 'hsc_percentage'] = median_hsc
```

```python
# For career_switch_willingness: if years_of_experience < 2, more likely willing

def impute_career_switch(row):
    if pd.isnull(row['career_switch_willingness']):
        if row['years_of_experience'] < 2:
            return 'Willing'
        else:
            return 'Not Willing'
    return row['career_switch_willingness']

df_clean['career_switch_willingness'] = df_clean.apply(impute_career_switch, axis=1)
```

```python
# For relevant_experience: if skills_match_percentage > 70%, likely relevant
def impute_relevant_experience(row):
    if pd.isnull(row['relevant_experience']):
        if row['skills_match_percentage'] > 70:
            return 'Relevant'
        else:
            return 'Not Relevant'
    return row['relevant_experience']

df_clean['relevant_experience'] = df_clean.apply(impute_relevant_experience, axis=1)
```

```python
# For job_role_match: if skills_match_percentage > 60%, likely matched
df_clean['job_role_match'].fillna(
    df_clean['skills_match_percentage'].apply(lambda x: 'Matched' if x > 60 else 'Not Matched'),
    inplace=True
)
```

```python
# For notice_period_days: fill with median based on experience
df_clean['notice_period_days'].fillna(
    df_clean.groupby('years_of_experience')['notice_period_days'].transform('median'),
    inplace=True
)
```

```python
# For layoff_history: if employment_gap_months > 6, likely Yes
def impute_layoff_history(row):
    if pd.isnull(row['layoff_history']):
        if row['employment_gap_months'] > 6:
            return 'Yes'
        else:
            return 'No'
    return row['layoff_history']

df_clean['layoff_history'] = df_clean.apply(impute_layoff_history, axis=1)
```

```python
# For employment_gap_months: fill with 0 for freshers
df_clean['employment_gap_months'].fillna(0, inplace=True)
```

```python
# For relocation_willingness: younger candidates more likely willing
def impute_relocation(row):
    if pd.isnull(row['relocation_willingness']):
        if row['age_years'] < 30:
            return 'Willing'
        else:
            return 'Not Willing'
    return row['relocation_willingness']

df_clean['relocation_willingness'] = df_clean.apply(impute_relocation, axis=1)
```

```python
# Verify no missing values remain
print("Remaining missing values:", df_clean.isnull().sum().sum())
```

```python
# Save cleaned data
df_clean.to_csv('../data/processed/cleaned_hr_data.csv', index=False)
```
